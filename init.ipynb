```python
input_path = 'seerah_transcripts.jsonl'
output_path = 'seerah_transcripts.jsonl'

with open(input_path, 'r', encoding='utf-8') as infile, \
     open(output_path, 'w', encoding='utf-8') as outfile:
    for line in infile:
        if not line.strip():
            continue
        # Only removes the specific \n characters
        cleaned_line = line.replace('\\n', '')
        outfile.write(cleaned_line)
```
